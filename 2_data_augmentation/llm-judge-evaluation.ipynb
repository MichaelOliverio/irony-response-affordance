```python
human_ranking_results = {
    'setting1': {
        'Coherence': [],
        'Engagingness': [],
        'Groundedness': [],
        'Naturalness': []
    },
    'setting2': {
        'Coherence': [],
        'Engagingness': [],
        'Groundedness': [],
        'Naturalness': []
    }
}

for annotator in human_annotators:
    for setting in human_annotators[annotator]:
        for dimension in human_annotators[annotator][setting]:
            score = human_annotators[annotator][setting][dimension]
            human_ranking_results[setting][dimension].append(score)

# Calcolo delle medie
average_human_ranking_results = {
    'setting1': {},
    'setting2': {}
}

for setting in human_ranking_results:
    for dimension in human_ranking_results[setting]:
        scores = human_ranking_results[setting][dimension]
        average_score = sum(scores) / len(scores)
        average_human_ranking_results[setting][dimension] = average_score

print("Average Human Ranking Results:")
for setting in average_human_ranking_results:
    print(f"{setting}:")
    for dimension in average_human_ranking_results[setting]:
        avg_score = average_human_ranking_results[setting][dimension]
        print(f"  {dimension}: {avg_score:.2f}")

# media tra le dimensioni
for setting in average_human_ranking_results:
    total = 0
    count = 0
    for dimension in average_human_ranking_results[setting]:
        total += average_human_ranking_results[setting][dimension]
        count += 1
    overall_average = total / count
    print(f"Overall Average for {setting}: {overall_average:.2f}")
```

```python
import pandas as pd
df_antonio = pd.read_excel("test_llm_judge_eval/Annotator_2_Antonio.xlsx", sheet_name="CSC")
df_antonio = df_antonio.head(32)
df_antonio = df_antonio.iloc[2:]

df_judge = pd.read_excel("test_llm_judge_eval/Annotator_2_Antonio.xlsx", sheet_name="CSC")
df_judge = df_judge.head(32)
df_judge = df_judge.iloc[2:]

# dalla colonna 3 alla 18 converti yes in 1 e il resto in 0
for col in df_antonio.columns[3:19]:
    df_antonio[col] = df_antonio[col].apply(lambda x: 1 if x == 'Yes' else 0)
for col in df_judge.columns[3:19]:
    df_judge[col] = df_judge[col].apply(lambda x: 1 if x == 'Yes' else 0)

df_judge.head()
```

```python
import json
with open("human_annotations/manual_eval_files/latin_square_assignments.json", "r") as f:
    data = json.load(f)

print(data['csc']['annotator_2']) # antonio
print(data['csc']['annotator_2']) # judge
```

```python
from collections import defaultdict

# estrai il dizionario dell'annotatore (antonio = annotator_2)
assignments = data['csc']['annotator_2']['unique_id_to_setting']

# crea dizionario che raggruppa gli id per setting
by_setting = defaultdict(list)

for unique_id, setting in assignments.items():
    by_setting[setting].append(unique_id)

# converti in dict normale (opzionale)
by_setting = dict(by_setting)

by_setting
```

```python
# Estrarre i mapping per annotator_2 (Antonio)
unique_id_to_setting_antonio = data['csc']['annotator_2']['unique_id_to_setting']
df_antonio['setting'] = df_antonio.iloc[:, 0].map(unique_id_to_setting_antonio) # Aggiungi la colonna 'setting' ai dataframe

unique_id_to_setting_judge = data['csc']['annotator_2']['unique_id_to_setting']
df_judge['setting'] = df_judge.iloc[:, 0].map(unique_id_to_setting_judge)

df_antonio.head()
```

```python
df_antonio['Coherence_total'] = df_antonio[['Coherence', 'Unnamed: 4', 'Unnamed: 5', 'Unnamed: 6']].sum(axis=1)
df_antonio = df_antonio.drop(columns=['Coherence', 'Unnamed: 4', 'Unnamed: 5', 'Unnamed: 6'])

df_antonio['Engagingness_total'] = df_antonio[['Engagingness', 'Unnamed: 8', 'Unnamed: 9']].sum(axis=1)
df_antonio = df_antonio.drop(columns=['Engagingness', 'Unnamed: 8', 'Unnamed: 9'])

df_antonio['Groundedness_total'] = df_antonio[['Groundedness', 'Unnamed: 11', 'Unnamed: 12', 'Unnamed: 13']].sum(axis=1)
df_antonio = df_antonio.drop(columns=['Groundedness',  'Unnamed: 11', 'Unnamed: 12', 'Unnamed: 13'])

df_antonio['Naturalness_total'] = df_antonio[['Naturalness', 'Unnamed: 15', 'Unnamed: 16', 'Unnamed: 17', 'Unnamed: 18']].sum(axis=1)
df_antonio = df_antonio.drop(columns=['Naturalness', 'Unnamed: 15', 'Unnamed: 16', 'Unnamed: 17', 'Unnamed: 18'])

df_antonio['Total_total'] = df_antonio[['Coherence_total', 'Engagingness_total', 'Groundedness_total', 'Naturalness_total']].sum(axis=1)

df_antonio.head()
```

```python
df_judge['Coherence_total'] = df_judge[['Coherence', 'Unnamed: 4', 'Unnamed: 5', 'Unnamed: 6']].sum(axis=1)
df_judge = df_judge.drop(columns=['Coherence', 'Unnamed: 4', 'Unnamed: 5', 'Unnamed: 6'])

df_judge['Engagingness_total'] = df_judge[['Engagingness', 'Unnamed: 8', 'Unnamed: 9']].sum(axis=1)
df_judge = df_judge.drop(columns=['Engagingness', 'Unnamed: 8', 'Unnamed: 9'])

df_judge['Groundedness_total'] = df_judge[['Groundedness', 'Unnamed: 11', 'Unnamed: 12', 'Unnamed: 13']].sum(axis=1)
df_judge = df_judge.drop(columns=['Groundedness',  'Unnamed: 11', 'Unnamed: 12', 'Unnamed: 13'])

df_judge['Naturalness_total'] = df_judge[['Naturalness', 'Unnamed: 15', 'Unnamed: 16', 'Unnamed: 17', 'Unnamed: 18']].sum(axis=1)
df_judge = df_judge.drop(columns=['Naturalness', 'Unnamed: 15', 'Unnamed: 16', 'Unnamed: 17', 'Unnamed: 18'])

df_judge['Total_total'] = df_judge[['Coherence_total', 'Engagingness_total', 'Groundedness_total', 'Naturalness_total']].sum(axis=1)

df_judge.head()
```

```python
df_merge = pd.merge(df_antonio, df_judge, on=['Unnamed: 0', 'setting'], suffixes=('_antonio', '_judge'))
df_merge.head()
```

```python
from scipy.stats import pearsonr, combine_pvalues
import pandas as pd

dimensions = ['Coherence', 'Engagingness', 'Groundedness', 'Naturalness', 'Total']

# lista per salvare i risultati
results = []

# loop su ogni setting
for setting, group in df_merge.groupby('setting'):
    row = {'setting': setting}
    for dim in dimensions:
        col_antonio = f"{dim}_total_antonio"
        col_judge = f"{dim}_total_judge"
        r, p = pearsonr(group[col_antonio], group[col_judge])
        row[f'{dim.lower()}_pearson_r'] = r
        row[f'{dim.lower()}_pearson_p'] = p
    results.append(row)

# crea dataframe finale
df_pearson = pd.DataFrame(results)
df_pearson
```
