```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

# Parametri Focal Loss
gamma = 0
alpha = 1

# Probabilità predetta per la classe corretta (da 0.01 a 0.99)
p_correct = torch.linspace(0.01, 0.99, 100)

# Funzione per convertire probabilità in logits
logits = torch.log(p_correct / (1 - p_correct)).unsqueeze(1)  # 2-class
logits_full = torch.cat([logits, -logits], dim=1)  # classi [correct, wrong]

labels = torch.zeros(len(p_correct), dtype=torch.long)  # classe corretta = 0

# CrossEntropy Loss senza media (per elemento)
loss_fct = nn.CrossEntropyLoss(reduction='none')
CE_loss = loss_fct(logits_full, labels)

# Focal Loss
pt = torch.exp(-CE_loss)
FL_loss = alpha * (1 - pt) ** gamma * CE_loss

# Plot
plt.figure(figsize=(8,5))
plt.plot(p_correct.numpy(), CE_loss.numpy(), label='CrossEntropy Loss', linewidth=2)
plt.plot(p_correct.numpy(), FL_loss.numpy(), label='Focal Loss', linewidth=2)
plt.xlabel("Probabilità predetta per la classe corretta")
plt.ylabel("Loss")
plt.title(f"Confronto CE vs Focal Loss (gamma={gamma}, alpha={alpha})")
plt.legend()
plt.grid(True)
plt.show()
```
