CSC Baseline

```python
import json

FILE_PATH = "CSC_meta-llama_Llama-3.1-8B-Instruct_n_turn_2_filtered_with_semiaggregates.json"

count_1 = 0
count_0 = 0

with open(FILE_PATH, "r", encoding="utf-8") as f:
    data = json.load(f)

for entry in data.get("entries", []):
    label = entry.get("annotations", {}).get("aggregate_sarcasm_label")
    if label == 1:
        count_1 += 1
    elif label == 0:
        count_0 += 1

total = count_1 + count_0
p = count_1 / total if total > 0 else 0

# F1 della baseline always-positive
f1_always_positive = (2 * p / (1 + p)) if p > 0 else 0

print("Distribuzione aggregate_sarcasm_label:")
print(f"Totale entry considerate: {total}")
print(f"Label = 1 : {count_1} ({p*100:.2f}%)")
print(f"Label = 0 : {count_0} ({(1-p)*100:.2f}%)")
print(f"F1 Always-Positive baseline: {f1_always_positive:.3f}")
```

```python
import json
import random
from sklearn.metrics import f1_score
import numpy as np

FILE_PATH = "CSC_meta-llama_Llama-3.1-8B-Instruct_n_turn_2_filtered_with_semiaggregates.json"
N_RUNS = 1000
SEED = 42

random.seed(SEED)
np.random.seed(SEED)

# Carica dati e ground truth
y_true = []

with open(FILE_PATH, "r", encoding="utf-8") as f:
    data = json.load(f)

for entry in data.get("entries", []):
    label = entry.get("annotations", {}).get("aggregate_sarcasm_label")
    if label in (0, 1):
        y_true.append(label)

y_true = np.array(y_true)

# Prevalenza positiva
p = y_true.mean()

f1_scores = []

for _ in range(N_RUNS):
    y_pred = np.random.binomial(1, p, size=len(y_true))
    f1 = f1_score(y_true, y_pred, pos_label=1)
    f1_scores.append(f1)

f1_scores = np.array(f1_scores)

print("Random(p) baseline")
print(f"Positive prevalence p: {p:.3f}")
print(f"F1 (mean ± std): {f1_scores.mean():.3f} ± {f1_scores.std():.3f}")
```
